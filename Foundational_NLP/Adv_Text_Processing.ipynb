# Text Normalization
Text Normalization is the process of cleaning and standardizing raw text to make it consistent and machine-readable. It handles issues like misspellings, contractions, emojis, HTML tags, URLs, hashtags, mentions, and special characters. Normalized text improves NLP model performance by reducing noise, standardizing vocabulary, and making feature extraction more effective.

```python
import re
import emoji
import contractions
from bs4 import BeautifulSoup
from textblob import TextBlob
```

**Sample Text**

```python
text = """
OMG!!! I can't believe this 😭😭
Check this out: https://example.com
@user said it's soooo amazingggg!!! #excited #NLP
<div>This is <b>HTML</b> content</div>
"""
```

```python
def adv_text_normalization(text):

  #   Remove HTML Tags
  text = BeautifulSoup(text, 'html.parser').get_text()

  # Expand Contractions
  text = contractions.fix(text)

  # Convert  Emojis to Text
  text = emoji.demojize(text, delimiters=(" ", " "))

  # Remove URLs
  text = re.sub(r'http\S+|www\S+', '', text)

  # Remove Mentions (@username)
  text = re.sub(r'@\w+', '', text)

  # Remove Hashtags (#tag → tag)
  text = re.sub(r'#(\w+)', r'\1', text)

  # Remove Special Characters
  text = re.sub(r'[^a-zA-Z\s]', '', text)

  # Lowercase
  text = text.lower()

  # Remove Extra Spaces
  text = re.sub(r'\s+', ' ', text).strip()

  # Spelling Correction
  text = str(TextBlob(text).correct())

  return text

```

```python
cleaned_text = adv_text_normalization(text)
print("Original Text: \n", text)
print("\n Cleaned Text: \n", cleaned_text)
```

# POS Tagging
POS Tagging is the process of identifying the grammatical role of each word in a sentence, such as nouns, verbs, adjectives, and adverbs. It helps NLP systems understand the structure and meaning of text.

```python
import nltk

# Download required NLTK data
nltk.download('punkt')  # Needed for word_tokenize
nltk.download('averaged_perceptron_tagger_eng')  # Needed for POS tagging

# Sample text
text = "Garima is building an amazing mental health app to help healthcare professionals."

# Tokenize the text into words
tokens = nltk.word_tokenize(text)

# Perform POS tagging
pos_tags = nltk.pos_tag(tokens)

# Print the result
print("POS Tags:")
for word, tag in pos_tags:
    print(f"{word} -> {tag}")
```

##  Named Entity Recognition
Named Entity Recognition (NER) is an NLP technique that automatically identifies and classifies key information (entities) in text into predefined categories such as:

PERSON – names of people

ORG – organizations or companies

LOCATION – cities, countries, places

DATE – dates or time expressions

It’s widely used in information extraction, research, chatbots, and text analysis to turn unstructured text into structured data.


```python
#  Install spaCy
!pip install spacy --quiet

# Download the English model
!python -m spacy download en_core_web_sm --quiet

# Import spaCy and load the model
import spacy

# Load the small English model
nlp = spacy.load("en_core_web_sm")

# Input text for NER
text = "Microsoft Research is in Bangalore. Satya Nadella works there since 2014."

doc = nlp(text)


print("Named Entities , Labels, and Spans: ")
for ent in doc.ents:
  print(f'{ent.text} -> {ent.label_}')

from spacy import displacy

displacy.render(doc, style='ent' , jupyter=True)

```

**Advance Name Entity Recognition**

```python
# Install Hugging Face Transformers and Torch
!pip install transformers torch --quiet



# Import libraries

from transformers import AutoTokenizer, AutoModelForTokenClassification, pipeline

model_name = "dbmdz/bert-large-cased-finetuned-conll03-english"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForTokenClassification.from_pretrained(model_name)

# Create a NER pipeline

ner_pipeline = pipeline("ner", model=model, tokenizer=tokenizer,aggregation_strategy="simple")

# Input text
text = 'Microsoft Research is in Bangalore. Satya Nadella works there since 2014.'

# Run NER
entities = ner_pipeline(text)

# print entities

print("Detected Entities: ")
for ent in entities:
  print(f"{ent['word']} -> {ent['entity_group']} (score: {ent['score']: .2f})")


```

## Dependency Parsing
Dependency parsing is an NLP technique that analyzes the grammatical structure of a sentence by establishing relationships between “head” words and their “dependents.”

Head word: The central word in a phrase (e.g., verb in a sentence).

Dependent: Words that modify or are related to the head (e.g., adjectives, objects).

Example sentence:
"Garima built a mental health app."

###  Transition-Based Dependency Parsing
Build the dependency tree incrementally using a stack and buffer.

Use actions like:

SHIFT → move next word from buffer to stack

LEFT-ARC → add dependency from next word to stack top, remove stack top





```python
# Simple Transition-Based Parser for educational purposes
class TransitionParser:
    def __init__(self, sentence):
        self.stack = ["ROOT"]
        self.buffer = sentence.split()
        self.arcs = []

    def parse(self):
        while self.buffer:
            # Simple rule: attach the first buffer word to top of stack
            self.arcs.append((self.stack[-1], self.buffer[0]))  # stack top -> buffer first
            # Move buffer word to stack
            self.stack.append(self.buffer.pop(0))
        return self.arcs

sentence = "Garima built a mental health app"
parser = TransitionParser(sentence)
arcs = parser.parse()
print("Dependency Arcs (head -> dependent):")
for head, dep in arcs:
    print(f"{head} -> {dep}")
```

### Graph-Based Dependency Parsing
Treat sentence as a graph: each word is a node.

All possible head-dependent edges are scored.

Use Maximum Spanning Tree (MST) algorithms (like Chu-Liu/Edmonds) to find the best tree

```python
import networkx as nx

sentence = ['Root', 'Garima', 'built', 'a', 'mental', 'health', 'app']

# Create an undirected weighted graph
G = nx.Graph()
for i, head in enumerate(sentence):
    for j, dep in enumerate(sentence):
        if i != j:
            # Assign a dummy weight (could be similarity or distance)
            G.add_edge(head, dep, weight=abs(i - j))

mst = nx.minimum_spanning_tree(G)

print("Dependency Arcs (head -> dependent):")
for head, dep in mst.edges():
    print(f"{head} -> {dep}")




```
