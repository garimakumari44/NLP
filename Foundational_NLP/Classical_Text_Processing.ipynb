```python
import nltk
import spacy
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_extraction.text import CountVectorizer


nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')


```

**Sample Text**

```python
text = """
At 2:17 a.m., the deployment pipeline failed because of a single misplaced environment variable.
The logs showed a silent timeout, not an obvious crash, which made debugging frustrating.
Meanwhile, the monitoring dashboard kept reporting green status indicators,
even though user requests were quietly queuing up and increasing latency.
In the background, a scheduled machine learning job continued retraining on incomplete data,
producing models that looked statistically valid but behaved unpredictably in production.
Technology rarely breaks dramatically; it usually degrades slowly,
hidden behind dashboards, metrics, and optimistic assumptions.
"""

```

# **Tokenization**
Tokenization is the process of breaking text into smaller units called tokens. These tokens can be words, sentences, or characters. It is the first step in Natural Language Processing and helps convert raw text into a format that machine learning models can understand.

```python
tokens = word_tokenize(text)
print("Tokens:\n", tokens)

```

# Lowercasing + Remove Punctuation
Lowercasing converts all text to lowercase so that words like "NLP", "nlp", and "Nlp" are treated as the same word. This reduces duplication in the vocabulary.

Removing punctuation eliminates symbols such as commas, periods, and question marks that usually do not add meaningful information for most NLP tasks.


```python
tokens = [word.lower() for word in tokens if word.isalpha() ]
print("Clean Tokens:\n", tokens)
```

# Stopword Removal
Stopword removal is the process of eliminating common words like “the”, “is”, “and”, and “in” from text.

These words appear frequently but usually do not carry significant meaning for many NLP tasks. Removing them helps reduce noise, decrease vocabulary size, and improve model efficiency.

```python
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word not in  stop_words]
print("After Stopword Removal:\n", filtered_tokens)
```

# Stemming
Stemming is the process of reducing words to their base or root form by removing suffixes.

For example, “running”, “runs”, and “runner” may all be reduced to “run”.

It helps reduce vocabulary size and improves text analysis by treating related words as the same term.

```python
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_tokens]
print("Stemmed Words: \n", stemmed_words)
```

# Lemmatization
Lemmatization is a text normalization technique in NLP that converts a word to its base or dictionary form (lemma) using vocabulary knowledge and morphological analysis.

Unlike stemming, it produces real, meaningful words

```python
lemmatizer = WordNetLemmatizer()
lemmatized_words = [ lemmatizer.lemmatize(word) for word in filtered_tokens ]
print("Lemmatized Words:\n", lemmatized_words)
```

## N-Grams (Unigram, Bigram)

N-grams are continuous sequences of N words from a sentence.

Unigram (1-gram) → Single words
Example:
“NLP is powerful” → ["NLP", "is", "powerful"]

Bigram (2-gram) → Pairs of consecutive words
Example:
“NLP is powerful” → ["NLP is", "is powerful"]

```python
vectorizer = CountVectorizer(ngram_range=(1,2))
X = vectorizer.fit_transform([text])

print("N-grams:\n", vectorizer.get_feature_names_out())
print('Matrix: \n', X.toarray())
```

## TF-IDF
TF-IDF (Term Frequency–Inverse Document Frequency) is a technique used in NLP to measure how important a word is in a document compared to a collection of documents.

TF (Term Frequency) → How often a word appears in a document.

IDF (Inverse Document Frequency) → How rare the word is across all documents.

TF-IDF gives:

Higher weight to important, unique words

Lower weight to very common words

```python
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform([text])

print("TF-DF Features: \n", tfidf.get_feature_names_out())
print("\nTF-IDF Matrix \n", tfidf_matrix.toarray())
```
