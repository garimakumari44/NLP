```python
import PyPDF2


def extract_text_from_pdf(pdf_path):
   text = ""
   with open(pdf_path, 'rb') as f:
      reader = PyPDF2.PdfReader(f)
      for page in reader.pages:
         text += page.extract_text() + "\n"
   return  text

pdf_path = "/content/EJ1172284.pdf"
paper_text = extract_text_from_pdf(pdf_path)
print(paper_text[:1000])
```

```python
def chunk_text(text, max_words=500):
    words = text.split()
    for i in range(0, len(words), max_words):
        yield " ".join(words[i:i+max_words])

chunks = list(chunk_text(paper_text, max_words=500))
print(f"Total chunks: {len(chunks)}")
```

```python
from transformers import BartTokenizer, BartForConditionalGeneration

# Load BART model
model_name = "facebook/bart-large-cnn"
tokenizer = BartTokenizer.from_pretrained(model_name)
model = BartForConditionalGeneration.from_pretrained(model_name)

summaries = []

for chunk in chunks:
    inputs = tokenizer([chunk], max_length=1024, return_tensors="pt", truncation=True)
    summary_ids = model.generate(
        inputs["input_ids"],
        num_beams=4,
        max_length=150,
        min_length=50,
        length_penalty=2.0,
        early_stopping=True
    )
    summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
    summaries.append(summary)

# Combine all chunk summaries into final summary
final_summary = " ".join(summaries)
print(final_summary)

```
